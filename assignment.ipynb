1.Explain the basic structure of an HTML page. Why is it important to follow
this structure?
 - `<DOCTYPE html>:`We start the file with `<!DOCTYPE html>`. It tells the browser what type of document to
expect; in this case it is an HTML document4
 - html lang="en":It's the language attribute and it tells the browser the default language. In this case english -en.
 - `<head>`:The <head> section contains metadata about the document, such as the page title, character encoding, viewport settings, and links to external resources like CSS stylesheets and JavaScript files.
 - `<title>`: The `<title>` element specifies the title of the document, which appears in the browser's title bar or tab and is crucial for search engine optimization.
 - `<body>`:The `<body>` section contains the visible content of the web page, including text, images, links, lists, videos, and other media.
 # Importance of HTML structure
 - HTML structure is important because it ensures a well-organized, maintainable, and accessible web page.

4.You’re creating a form that asks the user to enter their name, select their
country from a list, and click a button to submit. Write the HTML code for this.

```
<!DOCTYPE html>
<html lang="en">
<head>
<title> Hello, World!</title>
</head>
<body>
<form>
  <!-- Name input -->
  <label for="name">Name:</label><br>
  <input type="text" id="name" name="name"><br>
  <!-- Country selection -->
  <label for="country">Country:</label><br>
  <select id="country" name="country">
    <option value="usa">United States</option>
    <option value="canada">Canada</option>
    <option value="uk">United Kingdom</option>
    <option value="australia">Australia</option>
  </select><br>
  <!-- Submit button -->
  <button type="submit">Submit</button>
</form>
</body>
</html>
```

can see it, not the users on the website. How should he do this? Why are such notes useful?

```
<!-- This is a note for developers only -->

```
- Anything between `<!--` and `-->` is not displayed on the website, so users won’t see it, but developers can read it by viewing the source code.
# Why such notes are useful
- Explain code (what a section does or why it exists).
- Help team members understand and maintain the code.
- Make debugging easier by marking important sections.
- emind developers about future changes or TODOs.
